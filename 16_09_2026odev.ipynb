```python
# ==========================================
# 1. DEĞİŞKENLERLE BASİT HESAPLAMALAR
# ==========================================
birim_fiyat = 45.5
adet = 4
toplam_tutar = birim_fiyat * adet
print(f"Toplam Tutar: {toplam_tutar} TL")

maas = 8000
maas += 1500
print(f"Yeni Maaş: {maas} TL")


# ==========================================
# 2. İÇ İÇE (NESTED) IF-ELSE
# ==========================================
yas = int(input("Yaşınızı giriniz: "))
ogrenci_cevap = input("Öğrenci misiniz? (evet/hayir): ").strip().lower()

if yas > 65:
    print("Ücretsiz")
else:
    if ogrenci_cevap == "evet":
        print("İndirimli")
    else:
        print("Tam ücret")

# Sayı denetimi: Pozitif ve çift kontrolü
sayi = int(input("Bir tam sayı giriniz: "))
if sayi > 0:
    if sayi % 2 == 0:
        print("Sayı hem pozitif hem de çifttir.")
    else:
        print("Sayı pozitiftir ancak tektir.")
else:
    print("Sayı pozitif değildir.")


# ==========================================
# 3. ELIF İLE ÇOKLU KOŞULLAR
# ==========================================
sicaklik = float(input("Hava sıcaklığını giriniz (°C): "))
if sicaklik > 30:
    print("Sıcak")
elif 15 <= sicaklik <= 30:
    print("Ilıman")
else:
    print("Soğuk")

ay = int(input("Ay numarasını giriniz (1-12): "))
if ay in [12, 1, 2]:
    print("Kış")
elif ay in [3, 4, 5]:
    print("İlkbahar")
elif ay in [6, 7, 8]:
    print("Yaz")
elif ay in [9, 10, 11]:
    print("Sonbahar")
else:
    print("Geçersiz ay numarası!")


# ==========================================
# 4. FOR DÖNGÜSÜ İLE LİSTE İŞLEMLERİ
# ==========================================
sayilar = [5, 15, 8, 22, 3, 19]

# max() kullanmadan en büyük sayıyı bulma
en_buyuk = sayilar[0]
for s in sayilar:
    if s > en_buyuk:
        en_buyuk = s
print(f"En büyük sayı: {en_buyuk}")

# 10'dan büyükleri sayma
on_uzeri_sayac = 0
for s in sayilar:
    if s > 10:
        on_uzeri_sayac += 1
print(f"10'dan büyük sayı adedi: {on_uzeri_sayac}")


# ==========================================
# 5. WHILE DÖNGÜSÜ İLE BASİT SAYAÇ
# ==========================================
sayac = 10
while sayac >= 1:
    print(sayac)
    sayac -= 1

dogru_sifre = "python123"
girilen_sifre = ""
while girilen_sifre != dogru_sifre:
    girilen_sifre = input("Şifrenizi giriniz: ")
    if girilen_sifre != dogru_sifre:
        print("Hatalı şifre, tekrar deneyin.")
print("Giriş başarılı!")


# ==========================================
# 6. DÖNGÜ İÇİNDE KOŞUL KULLANMA
# ==========================================
print("1-20 arası 3'e bölünenler:")
for i in range(1, 21):
    if i % 3 == 0:
        print(i, end=" ")
print()

karisik_sayilar = [-10, 15, -3, 22, 0, -8, 7]
pozitifler = []
negatifler = []

for s in karisik_sayilar:
    if s > 0:
        pozitifler.append(s)
    elif s < 0:
        negatifler.append(s)

print(f"Pozitif Sayılar: {pozitifler}")
print(f"Negatif Sayılar: {negatifler}")


# ==========================================
# 7. BASİT FONKSİYON TANIMLAMA
# ==========================================
def kup_al(sayi):
    return sayi ** 3

def topla(a, b):
    return a + b

print("4'ün küpü:", kup_al(4))
print("12 + 25 =", topla(12, 25))


# ==========================================
# 8. FONKSİYON + KOŞUL BİRLİKTE
# ==========================================
def sayi_durumu(sayi):
    if sayi > 0:
        return "Pozitif"
    elif sayi < 0:
        return "Negatif"
    else:
        return "Sıfır"

def en_buyuk_bul(a, b, c):
    en_b = a
    if b > en_b:
        en_b = b
    if c > en_b:
        en_b = c
    return en_b

print("Durum (-5):", sayi_durumu(-5))
print("En büyük (14, 42, 28):", en_buyuk_bul(14, 42, 28))


# ==========================================
# 9. LİSTE OLUŞTURMA VE İŞLEME
# ==========================================
sehirler = []
for i in range(5):
    sehir_adi = input(f"{i+1}. şehri giriniz: ")
    sehirler.append(sehir_adi)

print("\n--- Şehir Listesi ---")
for sira, sehir in enumerate(sehirler, start=1):
    print(f"{sira}. {sehir}")

urunler = ["Laptop", "Klavye", "Mouse", "Kulaklık"]
print(f"\nMevcut ürünler: {urunler}")
silinecek = input("Silmek istediğiniz ürünü yazın: ")
if silinecek in urunler:
    urunler.remove(silinecek)
    print(f"Güncel liste: {urunler}")
else:
    print("Ürün listede bulunamadı.")


# ==========================================
# 10. KULLANICIDAN BİRDEN FAZLA VERİ ALMA
# ==========================================
ad = input("Adınız: ")
soyad = input("Soyadınız: ")
dogum_yili = int(input("Doğum yılınız: "))
yas_2026 = 2026 - dogum_yili
print(f"Sayın {ad} {soyad}, 2026 yılı itibarıyla yaşınız: {yas_2026}")

boy_cm = float(input("Boyunuzu (cm) giriniz: "))
kilo = float(input("Kilonuzu (kg) giriniz: "))
boy_m = boy_cm / 100
bmi = kilo / (boy_m ** 2)
print(f"Vücut Kitle İndeksiniz (BMI): {bmi:.2f}")


# ==========================================
# 11. UYGULAMA: NOT ORTALAMASI HESAPLAMA
# ==========================================
notlar = []
for i in range(4):
    puan = float(input(f"{i+1}. ders notunu giriniz: "))
    notlar.append(puan)

ortalama = sum(notlar) / len(notlar)

if ortalama >= 90:
    harf = "AA"
elif ortalama >= 85:
    harf = "BA"
elif ortalama >= 80:
    harf = "BB"
elif ortalama >= 70:
    harf = "CB"
elif ortalama >= 60:
    harf = "CC"
elif ortalama >= 50:
    harf = "DC"
else:
    harf = "FF"

print(f"\nOrtalama: {ortalama:.2f} | Harf Notu: {harf}")
print(f"En Yüksek Not: {max(notlar)}")
print(f"En Düşük Not: {min(notlar)}")


# ==========================================
# 12. UYGULAMA: BASİT SEPET
# ==========================================
sepet = [
    {"urun": "Defter", "fiyat": 60},
    {"urun": "Sırt Çantası", "fiyat": 350},
    {"urun": "Kulaklık", "fiyat": 220},
    {"urun": "Kalem", "fiyat": 25}
]

# 100 TL üzeri ürünleri filtreleme
pahali_urunler = [u["urun"] for u in sepet if u["fiyat"] > 100]
print(f"100 TL üzeri ürünler: {pahali_urunler}")

# Toplam sepet tutarı ve indirim kontrolü
toplam_sepet = sum(u["fiyat"] for u in sepet)
print(f"Ara Toplam: {toplam_sepet} TL")

if toplam_sepet > 500:
    odenecek_tutar = toplam_sepet * 0.90
    print(f"%10 indirim kazandınız! Ödenecek güncel tutar: {odenecek_tutar:.2f} TL")
else:
    print(f"Ödenecek tutar: {toplam_sepet} TL")
```
